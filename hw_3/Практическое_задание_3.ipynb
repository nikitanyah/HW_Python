Посчитайте процент выживаемости у каждого класса пассажиров (Pclass) - 1 балл

```python
survival = df.groupby("Pclass")["Survived"].mean()*100
survival.round(2)
```

Выведите самое популярное мужское и самое популярное женское имя на корабле - 1 балл

```python
def firstname(name):
    if '(' in name:
        start = name.find('(') + 1
        end = name.find(')')
        return name[start:end].split()[0]
    else:
        return name.split('.')[1].strip().split()[0]

df["FirstName"] = df["Name"].apply(firstname)
df.head()
```

```python
male_names = df[df["Sex"] == "male"]["FirstName"].value_counts().idxmax()
female_names = df[df["Sex"] == "female"]["FirstName"].value_counts().idxmax()
```

Выведите самое популярное мужское и самое популярное женское имя на корабле в каждом классе - 2 балла

```python
def findname(group):
    return group.value_counts().idxmax()

nameclass = (
    df.groupby(['Pclass', 'Sex'])['FirstName']
    .apply(findname)
)
nameclass
```

Выведите часть таблицы с пассажирами, возраст которых больше 44 лет - 1 балл

Выведите количества n-местных кабин (в которых было 2, 3, 4, ... человека) - 1 балл

```python
df = df[df['Cabin'].notna()]
cabins = df['Cabin'].str.split().explode()
cabin_counts = cabins.value_counts()
result = cabin_counts.value_counts().sort_index()
for key, val in result.items():
    if key >= 2:
        print(f"{key}-местных кабин: {val}")
```
